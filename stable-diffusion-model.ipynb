```python
import torch
from diffusers import StableDiffusionXLPipeline
from PIL import Image
from IPython.display import display
import ipywidgets as widgets

model_path = "/kaggle/input/stable-diffusion-xl/pytorch/base-1-0/1"

pipe = StableDiffusionXLPipeline.from_pretrained(
    model_path, 
    torch_dtype=torch.float16
).to("cuda")

# ===============================
# 4️⃣ Set up prompt input
# ===============================
prompt_box = widgets.Text(
    value='A magical forest at dawn, ultra-detailed, cinematic lighting, digital painting',
    placeholder='Type your prompt here',
    description='Prompt:',
    layout=widgets.Layout(width='800px')
)

generate_button = widgets.Button(description="Generate Image", button_style='success')

output = widgets.Output()

def generate_image(b):
    with output:
        output.clear_output()
        prompt = prompt_box.value
        negative_prompt = "blurry, low quality, text, watermark"
        print("Generating image... please wait")
        
        # Generate image
        image = pipe(prompt, negative_prompt=negative_prompt, guidance_scale=9.0).images[0]
      
        display(image)
        image.save("generated_image.png")
        print("Image saved as generated_image.png")

generate_button.on_click(generate_image)
# Display widgets
display(prompt_box, generate_button, output)
```
