```python
import gradio as gr

def chatbot(user_input):
    user_input = user_input.lower().strip()
    if "hi" in user_input or "hello" in user_input:
        return "Hello! How are you?"
    elif "how are you" in user_input:
        return "I'm fine, thank you!"
    elif "your name" in user_input:
        return "I am a demo chatbot."
    else:
        return "Thanks for your message!"

with gr.Blocks() as demo:
    gr.Markdown("## Simple Demo Chatbot")
    user_input = gr.Textbox(placeholder="Type something...")
    output_box = gr.Textbox(label="AI Reply")
    
    user_input.submit(chatbot, user_input, output_box)
    submit_btn = gr.Button("Send")
    submit_btn.click(chatbot, user_input, output_box)

demo.launch()
```
