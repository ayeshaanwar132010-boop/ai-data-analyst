> # **Inferential Statistics**


> **A one-sample t-test was conducted to compare the sample mean car year (2015.3) with the claimed mean (2019.3).Since the p-value is greater than 0.05, we fail to reject the null hypothesis.**


```python
import numpy as np
from scipy import stats

cars = np.array([
2018,2014,2017,2013,2019,2011,
2016,2015,2020,2010])

claimed_mean = 2019.3

n =  len(cars)
print("length of sample: ", n)

sample_mean = np.mean(cars)
print("Sample Mean:", sample_mean)

sample_std = np.std(cars, ddof=1)
print("Sample Std Dev:", sample_std)

SE = sample_std / np.sqrt(n)
print("Standard Error:", SE)

t_manual = (sample_mean - claimed_mean) / SE
print("Manual t-statistic:", t_manual)

df = n - 1
print("Degrees of Freedom:", df)

p_manual = 2 * (1 - stats.t.cdf(abs(t_manual), df=df))
print("Manual p-value:", p_manual)


t_statistic, p_value = stats.ttest_1samp(cars, claimed_mean)
print("\nSciPy t-statistic:", t_statistic)
print("SciPy p-value:", p_value)


print("Sample Mean:", np.mean(cars))
print("T-statistic:", t_statistic)
print("P-value:", p_value)

alpha = 0.05
if p_value < alpha:
    print("Reject the null hypothesis (H0).")
else:
    print("Fail to reject the null hypothesis (H0).")
```

> 
**The interquartile range (IQR) method was used to detect outliers in the car year data.All values fall within the lower and upper bounds, so no outliers were removed and the cleaned data is the same as the original.**


```python
import numpy as np
from scipy import stats


cars = np.array([
2018,2014,2017,2013,2019,2011,
2016,2015,2020,2010])


Q1 = np.percentile(cars, 25) 
Q3 = np.percentile(cars, 75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
cleaned_data = cars[(cars >= lower_bound) & (cars <= upper_bound)]

# Print results
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("\nOriginal Data:", cars)
print("Cleaned Data:", cleaned_data)
```
